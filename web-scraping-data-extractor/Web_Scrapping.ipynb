```python
import pandas as pd
import requests
import json
```

```python
api_url="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
```

```python
def get_number_of_jobs_T(technology):
    response = requests.get(api_url)
    data = response.json()

    number_of_jobs = 0
    
    for job in data:
        if technology.lower() in job['Key Skills'].lower():
            number_of_jobs += 1
    return technology,number_of_jobs

get_number_of_jobs_T("Python")
```

```python
#Write a function to find number of jobs in US for a location of your choice
def get_number_of_jobs_L(location):
    
    response = requests.get(api_url)
    data = response.json()

    number_of_jobs = 0

    for job in data:
        if location.lower() in job['Location'].lower():
            number_of_jobs += 1
    return location,number_of_jobs

get_number_of_jobs_L("Los Angeles")
```

```python
technologies = [
    "C", "C#", "C++", "Java", "JavaScript", "Python",
    "Scala", "Oracle", "SQL Server", "mysql",
    "PostgreSQL", "MongoDB"
]

results = []

for tech in technologies:
    results.append(get_number_of_jobs_T(tech))

# create dataframe
df = pd.DataFrame(results, columns=["Technology", "Number of job postings"])

# save to excel
df.to_excel("job-postings.xlsx", index=False)

print(df)
print("Excel file created successfully.")
```
